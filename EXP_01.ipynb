```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/ML_LAB/loan_prediction_dataset.csv")
df.head()
```

```python
df = pd.get_dummies(
    df,
    columns=['Employment_Status'],
    dtype=int
)
```

```python
df.head()
```

```python
df['Loan_Amount'] = (
    df['Loan_Amount'] - df['Loan_Amount'].min()
) / (
    df['Loan_Amount'].max() - df['Loan_Amount'].min()
)
```

```python
n = int(0.8 * len(df))

train = df[:n]
test = df[n:]


print("Training Data:")
print(train.head())

print("Testing Data:")
print(test.head())

print("Training Shape:", train.shape)
print("Testing Shape:", test.shape)
```
