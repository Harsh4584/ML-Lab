```python
import pandas as pd
import numpy as np

data = {
    'Age': [32, 45, 28, 50, 35, 40, 34],
    'Income': [55000, 70000, 40000, 80000, 60000, 75000, 56000],
    'Spend Frequency': [3, 2, 4, 1, 3, 2, 3],
    'Average Order Value': [200, 350, 150, 500, 250, 300, 275],
    'Valued Customer': ['Y', 'N', 'N', 'Y', 'Y', 'N', '?']
}

df = pd.DataFrame(data)
df
```

```python
X = df[['Age', 'Income', 'Spend Frequency', 'Average Order Value']].values

y = df['Valued Customer'].values

X_train = X[:-1]
y_train = y[:-1]

new_data = X[-1]

k = 3

distances = []

for i in range(len(X_train)):
    distance = np.sqrt(np.sum((X_train[i] - new_data) ** 2))
    distances.append((distance, y_train[i], i + 1))

distances.sort()

print("Nearest Neighbours:")

for i in range(k):
    print("Customer", distances[i][2],
          "Distance =", round(distances[i][0], 2),
          "Class =", distances[i][1])

neighbors = []

for i in range(k):
    neighbors.append(distances[i][1])

prediction = max(set(neighbors), key=neighbors.count)

print("Predicted Valued Customer:", prediction)
```
